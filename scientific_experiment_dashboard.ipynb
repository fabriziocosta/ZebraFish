# ZebraFish autonomous experiment dashboard

This notebook reads the scientific state and campaign artifacts without changing them. It shows the active run, recent trials, hypotheses, deterministic observations, candidate experiments, and a Graphviz reasoning graph with concise scientific node summaries.

Use `level=0` for a coarse project view and `level=5` for the most detailed evidence and graph labels.

```python
from pathlib import Path
import sys

def find_repo_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for candidate in (current, *current.parents):
        if (candidate / 'state' / 'scientific_state.yaml').exists() or (candidate / '.git').exists():
            return candidate
    return current

ROOT = find_repo_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

STATE_PATH = ROOT / 'state' / 'scientific_state.yaml'
CAMPAIGN_STATE_PATH = None  # Set to a campaign_state.json path to pin one campaign.
from src.scientific_dashboard import render_dashboard, dashboard_widget, build_dashboard_snapshot

print(f'Repository: {ROOT}')
print(f'Scientific state: {STATE_PATH}')
```

## Interactive granularity

The slider re-renders the same state at levels 0–5. Level 0 keeps only the main trial/question/hypothesis structure; higher levels add stages, observations, beliefs, candidates, metadata, relations, and edge labels. The graph only shows trials and stages that have deterministic observation evidence. Graph nodes identify the entity type, short ID, status, and the most useful score, question, hypothesis, or observation text. Trial nodes are numbered lexicographically, long text is wrapped, and numeric values are compacted to at most three significant digits or scientific notation. Hover over a node to see the complete untruncated text.

```python
dashboard_widget(
    state_path=STATE_PATH,
    campaign_state_path=CAMPAIGN_STATE_PATH,
)
```
